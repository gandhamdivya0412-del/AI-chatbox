```python
def rule_based_chatbot():
    print("Chatbot: Hello! I am a simple rule-based AI. (Type 'exit' or 'bye' to quit)\n")
    
    # Continuous loop to keep the conversation going
    while True:
        # Get user input and convert to lowercase for easier matching
        user_input = input("You: ").strip().lower()
        
        # 1. Exit Commands
        if user_input in ["exit", "bye", "quit", "goodbye"]:
            print("Chatbot: Goodbye! Have a great day!")
            break  # Exit the loop
        
        # 2. Greetings
        elif user_input in ["hello", "hi", "hey", "greetings"]:
            print("Chatbot: Hello there! How can I help you today?")
            
        # 3. Decision-Making Logic (Rule Sets)
        elif "how are you" in user_input:
            print("Chatbot: I'm just a computer program, but I'm functioning perfectly!")
            
        elif "your name" in user_input:
            print("Chatbot: I am PyBot, a basic rule-based AI assistant.")
            
        elif "help" in user_input:
            print("Chatbot: I can answer basic questions. Try asking about my name or how I am doing.")
            
        elif "weather" in user_input:
            print("Chatbot: I can't check live weather yet, but I hope it's sunny where you are!")
            
        # 4. Fallback response for unhandled inputs
        else:
            print("Chatbot: I'm sorry, I don't understand that yet. Try asking something else!")

# Run the chatbot
if __name__ == "__main__":
    rule_based_chatbot()
```
